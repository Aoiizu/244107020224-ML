```python
from sklearn.feature_extraction.text import CountVectorizer

df = pd.read_csv(DATA + 'spam.csv', encoding='latin-1')
df = df.iloc[:, :2]
df.columns = ['Labels', 'SMS']
df.head()
```

```python
print(df['Labels'].value_counts()); print()
print(df.info()); print()
print(df.describe())
```

```python
df['Labels'] = df['Labels'].map({'spam': 1, 'ham': 0})
X = df['SMS'].values
y = df['Labels'].values
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=50)
bow = CountVectorizer()
X_train_bow = bow.fit_transform(X_train)
X_test_bow = bow.transform(X_test)
print(len(bow.get_feature_names_out()))
print(f'Data dimensions: {X_train_bow.shape}')
```

```python
mnb = MultinomialNB().fit(X_train_bow, y_train)
print(f'Training data accuracy: {accuracy_score(y_train, mnb.predict(X_train_bow))}')
print(f'Test data accuracy: {accuracy_score(y_test, mnb.predict(X_test_bow))}')
```
